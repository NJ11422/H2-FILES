**Task 3.1**

Write a Python program that uses SQL to create a database with the **three** tables: `User`, `Post` and `Comment`

Define the primary and foreign keys for each table. **[6]**


````python
import sqlite3
connection = sqlite3.connect("media.db")
connection.execute("DROP TABLE IF EXISTS User")
connection.execute("DROP TABLE IF EXISTS Post")
connection.execute("DROP TABLE IF EXISTS Comment")

connection.execute("""CREATE TABLE User(
                  Username TEXT NOT NULL,
                  Blocked INTEGER NOT NULL,
                  PRIMARY KEY(Username))
                  """)

connection.execute("""CREATE TABLE Post (
                  PostID INTEGER NOT NULL,
                  PostText TEXT NOT NULL,
                  DateTimePosted TEXT NOT NULL,
                  Username TEXT NOT NULL,
                  PRIMARY KEY(PostID),
                  FOREIGN KEY(Username) REFERENCES User(Username))
                  """)

connection.execute("""CREATE TABLE Comment (
                  CommentID INTEGER NOT NULL,
                  PostID INTEGER NOT NULL,
                  Username TEXT NOT NULL,
                  TheText TEXT NOT NULL,
                  DateTimeCommented TEXT NOT NULL,
                  PRIMARY KEY(CommentID),
                  FOREIGN KEY(PostID) REFERENCES Post(PostID),
                  FOREIGN KEY(Username) REFERENCES User(Username))
                  """)

````

**Task 3.2**

Write a Python program that uses SQL to insert the data from each of the **three** text files into the appropriate fields in each database table. **[4]**

Run your program. **[1]**


````python
#Task 3.2
import sqlite3
connection = sqlite3.connect("media.db")
openfile = open("user.txt")
data = openfile.readlines()
for i in data:
  i = i.strip()
  User, Block = i.split(',')
  connection.execute("INSERT INTO User(Username, Blocked)VALUES(?,?)",(User, Block))


with open("post.txt", encoding="utf-8-sig") as f:
    data = f.readlines()
for i in data:
  i = i.strip()
  PostID, PostText, DateTimePosted, Username = i.split(',')
  connection.execute("""INSERT INTO Post(PostID, PostText, DateTimePosted, Username)VALUES(?,?,?,?)""",(PostID, PostText, DateTimePosted, Username))

with open("comment.txt", encoding="utf-8-sig") as f:
    data = f.readlines()
for i in data:
  i = i.strip()
  CommentID, PostID, Username, TheText, DateTimeCommented = i.split(',')
  connection.execute("""INSERT INTO Comment(CommentID, PostID, Username, TheText, DateTimeCommented)VALUES(?,?,?,?,?)""",(CommentID, PostID, Username, TheText, DateTimeCommented))

````

**Task 3.3**

Write a Python function that:

- uses SQL to search for all comments by a user who is blocked
- outputs each returned username and comment on a new line.

**[5]**

Run your program. **[1]**

Save your Jupyter Notebook for Task 3.


````python
#Task 3.3
import sqlite3
connection = sqlite3.connect("media.db")
data = connection.execute("""SELECT Username FROM User WHERE Blocked = 1""")
for documents in data:
    values = connection.execute("""SELECT TheText FROM Comment WHERE Username = ?""",(documents))
    for text in values:
        print(documents[0] + "'s comment: " + text[0])
connection.commit()


````
