**Task 3.1**

Write the `LinkedList` class in Python. Use of a simple Python list is not sufficient. Include the following methods:

- `insert(integer_value)` inserts the `integer_value` at the beginning (head) of the list
- `delete(integer_value)` attempts to delete `integer_value` from the list; if the item was not present, return `None`
- `search(integer_value)` returns a Boolean value: `True` if `integer_value` is in the list, `False` if not in the list
- `count()` should return the number of elements in the list, or zero if empty
- `to_String()` should return a string containing a suitably formatted list with the elements separated by a comma and a space, with square brackets at either end, e.g. in the form:

  `[11, 2, 7, 4]`

  **[8]**

Test `LinkedList` by using the data in the file `Task3data.txt`. Use the `to_String()` method to print the resulting contents of the list. **[3]**


````python
# Task 3.1
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class LinkedList:
  def __init__(self):
    self.head = None

  def isEmpty(self):
    if self.head == None:
      return True
    else:
      return False

  def insert(self, integer_value):
    newnode = Node(integer_value)

    if self.isEmpty():
      self.head =  newnode

    else:

      newnode.next = self.head

      self.head = newnode


  def delete(self, integer_value):
    if self.isEmpty():
      return None

    current = self.head
    prev = None
    while current != None and current.data != integer_value:
      prev = current
      current = current.next

    if prev == None:
      self.head = current.next
      return True

    if current.data == integer_value:
      prev.next = current.next
      return True

    return None

  def search(self,integer_value):
    if self.isEmpty():
      return False

    current = self.head
    prev = None
    while current != None and current.data != integer_value:
      prev = current
      current = current.next

    if current.data == integer_value:
      return True

    return False

  def count(self):
    count = 0
    if self.isEmpty():
      return False

    current = self.head
    prev = None
    while current != None:
      prev = current
      current = current.next
      count = count + 1

    return count

  def to_String(self):
    lst = []
    if self.isEmpty():
      return False

    current = self.head
    prev = None
    while current != None:
      lst.append(current.data)
      prev = current
      current = current.next

    return lst

person = LinkedList()
openfile = open("Task3data.txt")
data = openfile.readlines()
for i in data:
  i = i.strip()
  i = int(i)
  person.insert(i)

person.to_String()


````

Write a Python subclass `SortedLinkedList` using `LinkedList` as its superclass.

````python
# Task 3.2
class SortedLinkedList(LinkedList):
  def __init__ (self):
    super().__init__()
    return

  def insert(self, integer_value):
    newnode = Node(integer_value)

    if self.isEmpty():
      self.head =  newnode
      return True

    else:
      current = self.head
      prev = None
      while current != None and integer_value >= current.data :
        prev = current
        current = current.next

      if prev == None:
        newnode.next = self.head
        self.head = newnode
        return True

      prev.next = newnode
      newnode.next = current
      return True



person = SortedLinkedList()
openfile = open("Task3data.txt")
data = openfile.readlines()
for i in data:
  i = i.strip()
  i = int(i)
  person.insert(i)
person.to_String()


````
